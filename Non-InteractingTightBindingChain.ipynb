```python
N = 50 # length of chain 
t = 1 # hopping term 
Nkeep = 300 
tol = Nkeep * 100 * 10 **(-16)
F = np.array([[0,1],[0,0]]) # annihilation operator 
Z = np.array([[1, 0],[0, -1]]) # Zordan wigner string operator 
I = np.eye(2)
```

```python
N = 50
E_g = [] # save the ground energy eigenstate 

Hprev = np.array([0]).reshape((1,1)) # Hamiltonian of vacuum 
Aprev = np.array([1]).reshape((1,1,1)) # identity tensor for the dummy leg 

for j in range(N):
    Anow = get_identity(Aprev, 1, I, 1)
    Hnow = updateLeft(Hprev, Anow, None, Anow)
    
    if j >0: 
        # add hoping terms 
        ZF = contract(Z, F, [1], [0])
        Hhop = - t * updateLeft(Fprev, Anow, np.transpose(np.conj(ZF), (1,0)), Anow)
        Hnow = Hnow + Hhop + Hhop.T.conj() 
        
    
    D, V = np.linalg.eigh((Hnow+Hnow.T.conj())/2)
    if np.mean(np.abs(Hnow - (Hnow+Hnow.T.conj())/2))>10**(-14):
        print("Hamiltonian is not hermitian")
    E_g.append(D[0])
    
    # truncate threshold for energy 
    E_threshold = D[min(len(D), Nkeep)-1]
    oks=D<(E_threshold+tol)
    
    # truncate 
    V = V[:, oks]
    Hprev = np.diag(D[oks])
    
    Aprev = contract(Anow, V, [1], [0])
    Aprev = np.transpose(Aprev, (0, 2, 1)) # left, right, bottom 
    Fprev = updateLeft(None, Aprev, F, Aprev)
    
```
