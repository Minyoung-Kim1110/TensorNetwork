```python
J = 1 # interaction strength 
N = 3 # number of spins (triangle)
hbar = 1 


# for spin 1/2 
I = np.eye(2)
sigma_x = np.array([[0, 1],[1, 0]])
sigma_y = np.array([[0, 0-1j],[0+1j, 0]])
sigma_z = np.array([[1, 0],[0, -1]])

S = np.array([sigma_x, sigma_y, sigma_z])*hbar/2 #or we can set up with Sigma_+, Sigma_-,Sigma_z, 
S = np.transpose(S, (1,2,0))
```

```python
Aprev = np.array([1]).reshape((1,1,1))
H = np.array([0]).reshape((1,1))
SpinOperators = dict() 

for i in range(N):
    Anow = get_identity(Aprev, len(Aprev.shape)-2, I, 1)
    # print('A', Anow.shape)
    # Contract Hamiltonian up to the last iteration
    H = update_left(H, Anow, None, Anow)
    # print('Hamiltonian up to ', i+1, 'th particle:', H.shape)
    
    for j in range(i):
        Hsp = update_left(SpinOperators[j], Anow, np.transpose(np.conj(S), (1, 0, 2)), Anow)
        # print("HSP", Hsp.shape)
        H = H+J*Hsp 
        
    
    for j in range(i+1):
        if j<i: 
            # print("previous", SpinOperators[j].shape)
            SpinOperators[j]=update_left(SpinOperators[j], B=Anow, X=None, A=Anow)
            # print("after", SpinOperators[j].shape)
            
        else: 
            SpinOperators[j] = update_left(None, B = Anow, X = S, A = Anow)
            # print('last', SpinOperators[j].shape)
    Aprev = Anow
```

```python
# H is Hermitial matrix but to erase the noises, 
H = (H + H.conj().T)/2

D, V = np.linalg.eigh(H)
```
